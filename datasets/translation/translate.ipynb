```python
import pandas as pd
from googletrans import Translator

# ---- Load your CSV file ----
file_path = "../cleanedDataset.csv"
df = pd.read_csv(file_path)

translator = Translator()

# Lists to store translations
english_names = []
english_desc = []
marathi_names = []
marathi_desc = []

# ---- Translation loop ----
for name, desc in zip(df["scheme_name"], df["description"]):
    # Translate to English
    en_name = translator.translate(str(name), dest="en").text
    en_desc = translator.translate(str(desc), dest="en").text

    # Translate to Marathi
    mr_name = translator.translate(str(name), dest="mr").text
    mr_desc = translator.translate(str(desc), dest="mr").text

    english_names.append(en_name)
    english_desc.append(en_desc)
    marathi_names.append(mr_name)
    marathi_desc.append(mr_desc)

# ---- Create English dataset ----
df_english = df.copy()
df_english["scheme_name"] = english_names
df_english["description"] = english_desc

# ---- Create Marathi dataset ----
df_marathi = df.copy()
df_marathi["scheme_name"] = marathi_names
df_marathi["description"] = marathi_desc

# ---- Save the new datasets ----
df_english.to_csv("english.csv", index=False)
df_marathi.to_csv("marathi.csv", index=False)

print("✅ Translation completed!")
print("English file saved as: english.csv")
print("Marathi file saved as: marathi.csv")
```

```python
import pandas as pd
from googletrans import Translator

# ---- Load your CSV file ----
file_path = "../cleanedDataset.csv"
df = pd.read_csv(file_path)

translator = Translator()

# Lists to store translations
english_names = []
english_desc = []
marathi_names = []
marathi_desc = []

def safe_translate(text, target):
    try:
        return translator.translate(str(text), dest=target).text
    except:
        return text

# ---- Translation loop ----
for name, desc in zip(df["scheme_name"], df["description"]):

    # Step 1: Force everything to Marathi first
    forced_mr_name = safe_translate(name, "mr")
    forced_mr_desc = safe_translate(desc, "mr")

    # Step 2: Create clean Marathi dataset
    mr_name = forced_mr_name
    mr_desc = forced_mr_desc

    # Step 3: Convert clean Marathi to pure English
    en_name = safe_translate(forced_mr_name, "en")
    en_desc = safe_translate(forced_mr_desc, "en")

    # Store results
    marathi_names.append(mr_name)
    marathi_desc.append(mr_desc)
    english_names.append(en_name)
    english_desc.append(en_desc)

# ---- Create Marathi dataset ----
df_marathi = df.copy()
df_marathi["scheme_name"] = marathi_names
df_marathi["description"] = marathi_desc

# ---- Create English dataset ----
df_english = df.copy()
df_english["scheme_name"] = english_names
df_english["description"] = english_desc

# ---- Save the new datasets ----
df_marathi.to_csv("pure_marathi.csv", index=False)
df_english.to_csv("pure_english.csv", index=False)

print("✅ Translation completed!")
print("Marathi file saved as: pure_marathi.csv")
print("English file saved as: pure_english.csv")
```
